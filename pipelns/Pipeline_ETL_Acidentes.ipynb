```python
import pandas as pd
import datetime
import locale

df_2014 = pd.read_csv('acidentes_2014.csv', sep=';', encoding='utf-8')
df_2015 = pd.read_csv('acidentes_2015.csv', sep=';', encoding='utf-8')
df_2016 = pd.read_csv('acidentes_2016.csv', sep=';', encoding='utf-8')

# Carregando dados mensais de 2015
df_janeiro_2015 = pd.read_csv('acidentes_janeiro_2015.csv', sep=',', encoding='utf-8')
df_fevereiro_2015 = pd.read_csv('acidentes_fevereiro_2015.csv', sep=',', encoding='utf-8')
df_marco_2015 = pd.read_csv('acidentes_marco_2015.csv', sep=',', encoding='utf-8')
df_abril_2015 = pd.read_csv('acidentes_abril_2015.csv', sep=',', encoding='utf-8')
df_maio_2015 = pd.read_csv('acidentes_maio_2015.csv', sep=',', encoding='utf-8')

print("Bases carregadas com sucesso!")
```

```python
# 2. TRANSFORMAÇÃO: Renomeando as colunas para o novo padrão semântico

map_2014 = {
    'tipo': 'tipo_acidente',
    'detalhes': 'veiculo'
}

map_2015 = {
    'data_abertura': 'data',
    'hora_abertura': 'hora',
    'tipo_ocorrencia': 'tipo_acidente',
    'tipo': 'veiculo'
}

map_2016 = {
    'data de abertura': 'data',
    'hora de abertura': 'hora',
    'tipo de ocorrencia': 'tipo_acidente',
    'tipo': 'veiculo'
}

# Mapeamento para dados mensais de 2015
map_mensal_2015 = {
    'tipo': 'tipo_acidente',
    'detalhes': 'veiculo'
}

colunas_desejadas = ['tipo_acidente', 'data', 'veiculo', 'longitude', 'latitude']

# Função de limpeza que é "à prova de erros"
def processar_arquivo_mensal(df):
    df.columns = df.columns.str.lower().str.strip() # Padroniza nomes
    df = df.rename(columns=map_mensal_2015)
    # Mantém apenas as colunas que existem na lista e que também estão no DF
    colunas_presentes = [c for c in colunas_desejadas if c in df.columns]
    return df[colunas_presentes]

# Aplicando as renomeações
df_2014 = df_2014.rename(columns=map_2014)
df_2015 = df_2015.rename(columns=map_2015)
df_2016 = df_2016.rename(columns=map_2016)

df_janeiro_2015 = processar_arquivo_mensal(pd.read_csv('acidentes_janeiro_2015.csv', sep=';'))
df_fevereiro_2015 = processar_arquivo_mensal(pd.read_csv('acidentes_fevereiro_2015.csv', sep=';'))
df_marco_2015 = processar_arquivo_mensal(pd.read_csv('acidentes_marco_2015.csv', sep=';'))
df_abril_2015 = processar_arquivo_mensal(pd.read_csv('acidentes_abril_2015.csv', sep=';'))
df_maio_2015 = processar_arquivo_mensal(pd.read_csv('acidentes_maio_2015.csv', sep=';'))

# Junta novamente os dados com os novos nomes
df_consolidado = pd.concat([df_2014, df_2015, df_2016, df_janeiro_2015, df_fevereiro_2015,
                             df_marco_2015, df_abril_2015, df_maio_2015], ignore_index=True)
print("Colunas renomeadas e dados consolidados com sucesso!")
```

```python

# Consolidando tudo em um único DataFrame
df_consolidado = pd.concat([df_2014, df_2015, df_2016], ignore_index=True)

# Visualizando o tamanho da nova base e as primeiras linhas
print(f"Total de registros consolidados: {df_consolidado.shape[0]} linhas e {df_consolidado.shape[1]} colunas")
display(df_consolidado.head())
```

```python
# ==========================================
# 3. TRANSFORMAÇÃO: Limpeza e Tipagem de Dados (ATUALIZADO)
# ==========================================

# 1. Tratando os valores nulos (NaN)
colunas_texto = ['hora', 'bairro', 'endereco', 'complemento', 'descricao']
df_consolidado[colunas_texto] = df_consolidado[colunas_texto].fillna('Não Informado')
df_consolidado['quantidade_vitimas'] = df_consolidado['quantidade_vitimas'].fillna(0)

# 2. Corrigindo a Tipagem da Data
df_consolidado['data'] = pd.to_datetime(df_consolidado['data'], errors='coerce')

# 3. Padronização de Texto (Tudo em maiúsculo)
df_consolidado['tipo_acidente'] = df_consolidado['tipo_acidente'].str.upper().str.strip()
df_consolidado['veiculo'] = df_consolidado['veiculo'].str.upper().str.strip()
df_consolidado['bairro'] = df_consolidado['bairro'].str.upper().str.strip()

print("Limpeza concluída com as novas colunas!")
```

```python
# ==========================================
# 4. MODELAGEM DIMENSIONAL: Novas Tabelas Dimensão
# ==========================================

# 1. Dimensão Tipo Acidente
dim_tipo_acidente = df_consolidado[['tipo_acidente']].drop_duplicates().reset_index(drop=True)
dim_tipo_acidente.insert(0, 'sk_tipo_acidente', dim_tipo_acidente.index + 1)

# 2. Dimensão Veículo
dim_veiculo = df_consolidado[['veiculo']].drop_duplicates().reset_index(drop=True)
dim_veiculo.insert(0, 'sk_veiculo', dim_veiculo.index + 1)

# 3. Dimensão Local
dim_local = df_consolidado[['bairro', 'endereco', 'complemento']].drop_duplicates().reset_index(drop=True)
dim_local.insert(0, 'sk_local', dim_local.index + 1)

# 4. Dimensão Tempo
dim_tempo = df_consolidado[['data', 'hora']].drop_duplicates().reset_index(drop=True)
dim_tempo.insert(0, 'sk_tempo', dim_tempo.index + 1)
dim_tempo['ano'] = dim_tempo['data'].dt.year
dim_tempo['mes'] = dim_tempo['data'].dt.month
dim_tempo['dia'] = dim_tempo['data'].dt.day
dim_tempo['nome_dia_semana'] = dim_tempo['data'].dt.day_name()
dim_tempo['is_dia_semana'] = dim_tempo['data'].dt.weekday < 5
print("Novas Tabelas Dimensão criadas com sucesso!")
display(dim_tempo.head())
```

```python
# ==========================================
# 5. MODELAGEM DIMENSIONAL: Tabela Fato (CORRIGIDO)
# ==========================================

# Fazemos uma cópia da base limpa
fato_acidente = df_consolidado.copy()

# 1. Trazendo o ID da Dimensão Tipo Acidente (Note o nome da coluna correto)
fato_acidente = fato_acidente.merge(dim_tipo_acidente, on=['tipo_acidente'], how='left')

# 2. Trazendo o ID da Dimensão Veículo (Note o nome da coluna correto)
fato_acidente = fato_acidente.merge(dim_veiculo, on=['veiculo'], how='left')

# 3. Trazendo o ID da Dimensão Local
fato_acidente = fato_acidente.merge(dim_local, on=['bairro', 'endereco', 'complemento'], how='left')

# 4. Trazendo o ID da Dimensão Tempo
fato_acidente = fato_acidente.merge(dim_tempo[['data', 'hora', 'sk_tempo']], on=['data', 'hora'], how='left')

# Agora removemos as colunas de texto que já foram substituídas pelos IDs (SKs)
colunas_para_remover = [
    'tipo_acidente', 'veiculo',
    'bairro', 'endereco', 'complemento',
    'data', 'hora'
]
fato_acidente = fato_acidente.drop(columns=colunas_para_remover, errors='ignore')

# Criando a Chave Primária da Tabela Fato
fato_acidente.insert(0, 'sk_acidente', fato_acidente.index + 1)

# Reorganizando a ordem das colunas para ficar elegante
fato_acidente = fato_acidente[[
    'sk_acidente', 'sk_tempo', 'sk_local', 'sk_tipo_acidente', 'sk_veiculo',
    'quantidade_vitimas', 'latitude', 'longitude', 'descricao'
]]

print("Tabela Fato criada com sucesso!")
display(fato_acidente.head())
```

```python
# ==========================================
# 6. CARGA (LOAD): Exportando os Dados
# ==========================================

# Salvando a Tabela Fato
fato_acidente.to_csv('fato_acidente.csv', index=False, sep=';', encoding='utf-8')

# Salvando as Tabelas Dimensão
dim_ocorrencia.to_csv('dim_ocorrencia.csv', index=False, sep=';', encoding='utf-8')
dim_tipo_envolvido.to_csv('dim_tipo_envolvido.csv', index=False, sep=';', encoding='utf-8')
dim_local.to_csv('dim_local.csv', index=False, sep=';', encoding='utf-8')
dim_tempo.to_csv('dim_tempo.csv', index=False, sep=';', encoding='utf-8')

print("Processo ETL Finalizado! Arquivos CSV gerados e prontos para download.")
```

```python
import sqlite3
from sqlite3 import Error

def criar_tabelas_sqlite():
    # Nome do arquivo do banco de dados local
    # Se o arquivo não existir, o SQLite criará automaticamente na mesma pasta do script
    nome_banco = "dw_acidentes.db"

    # Scripts DDL para criação das tabelas
    # A ORDEM IMPORTA: Dimensões primeiro, Fato por último por causa das Foreign Keys.
    tabelas_sql = {
        "dim_local": """
            CREATE TABLE IF NOT EXISTS dim_local (
                sk_local INTEGER PRIMARY KEY,
                bairro TEXT,
                endereco TEXT,
                latitude REAL,
                longitude REAL
            );
        """,
        "dim_tempo": """
            CREATE TABLE IF NOT EXISTS dim_tempo (
                sk_tempo INTEGER PRIMARY KEY,
                data_acidente TEXT, -- SQLite armazena datas como TEXT (YYYY-MM-DD)
                ano INTEGER,
                mes INTEGER,
                dia INTEGER,
                hora_acidente TEXT, -- SQLite não tem tipo TIME nativo, usa-se TEXT (HH:MM:SS)
                dia_semana TEXT,
                fim_de_semana TEXT
            );
        """,
        "dim_tipo_acidente": """
            CREATE TABLE IF NOT EXISTS dim_tipo_acidente (
                sk_tipo_acidente INTEGER PRIMARY KEY,
                causa_acidente TEXT,
                descricao_detalhada TEXT
            );
        """,
        "dim_ocorrencia": """
            CREATE TABLE IF NOT EXISTS dim_ocorrencia (
                sk_ocorrencia INTEGER PRIMARY KEY,
                tipo_ocorrencia TEXT,
                classificacao_ocorrencia TEXT
            );
        """,
        "dim_veiculo": """
            CREATE TABLE IF NOT EXISTS dim_veiculo (
                sk_veiculo INTEGER PRIMARY KEY,
                tipo_veiculo TEXT,
                categoria_veiculo TEXT
            );
        """,
        "fato_acidente": """
            CREATE TABLE IF NOT EXISTS fato_acidente (
                sk_acidente INTEGER PRIMARY KEY,
                sk_tempo INTEGER,
                sk_local INTEGER,
                sk_tipo_acidente INTEGER,
                sk_ocorrencia INTEGER,
                sk_veiculo INTEGER,
                quantidade_vitimas INTEGER,
                quantidade_veiculos INTEGER,
                FOREIGN KEY (sk_tempo) REFERENCES dim_tempo(sk_tempo),
                FOREIGN KEY (sk_local) REFERENCES dim_local(sk_local),
                FOREIGN KEY (sk_tipo_acidente) REFERENCES dim_tipo_acidente(sk_tipo_acidente),
                FOREIGN KEY (sk_ocorrencia) REFERENCES dim_ocorrencia(sk_ocorrencia),
                FOREIGN KEY (sk_veiculo) REFERENCES dim_veiculo(sk_veiculo)
            );
        """
    }

    try:
        print(f"Conectando/Criando o banco de dados local '{nome_banco}'...")
        # Abre a conexão com o arquivo
        conexao = sqlite3.connect(nome_banco)
        cursor = conexao.cursor()

        # IMPORTANTE: Por padrão, o SQLite desativa a checagem de Foreign Keys.
        # Este comando ativa o suporte a chaves estrangeiras na sessão atual.
        cursor.execute("PRAGMA foreign_keys = ON;")

        print("Conexão bem-sucedida! Criando tabelas...\n")

        # Iterando sobre o dicionário e executando os comandos SQL
        for nome_tabela, ddl in tabelas_sql.items():
            print(f"Criando a tabela: {nome_tabela}...")
            cursor.execute(ddl)
            print(f"Tabela {nome_tabela} configurada com sucesso!\n")

        # Salva as alterações no arquivo
        conexao.commit()
        print("Processo finalizado com sucesso. O arquivo de banco de dados está pronto para uso.")

    except Error as e:
        print(f"Erro ao acessar o SQLite: {e}")
    finally:
        # Fecha a conexão com o arquivo de forma segura
        if conexao:
            cursor.close()
            conexao.close()
            print("Conexão com o SQLite encerrada.")

if __name__ == "__main__":
    criar_tabelas_sqlite()
```

```python
import sqlite3

# 1. Criar ou abrir a conexão com o mesmo banco de dados SQLite
nome_banco = "dw_acidentes.db"
conexao = sqlite3.connect(nome_banco)

try:
    print("Iniciando a carga dos dados no SQLite...\n")

    # =========================================================================
    # 2. Ajustar e Inserir: dim_tipo_acidente
    # =========================================================================
    # O banco espera 'causa_acidente'. Vamos renomear temporariamente para salvar.
    df_tipo_ajustado = dim_tipo_acidente.rename(columns={'tipo_acidente': 'causa_acidente'})
    # Criamos uma coluna vazia para a 'descricao_detalhada' caso não exista no seu DF
    if 'descricao_detalhada' not in df_tipo_ajustado.columns:
        df_tipo_ajustado['descricao_detalhada'] = None

    df_tipo_ajustado.to_sql('dim_tipo_acidente', con=conexao, if_exists='append', index=False)
    print("✓ Dados da 'dim_tipo_acidente' inseridos com sucesso!")

    # =========================================================================
    # 3. Ajustar e Inserir: dim_veiculo
    # =========================================================================
    df_veiculo_ajustado = dim_veiculo.rename(columns={'veiculo': 'tipo_veiculo'})
    if 'categoria_veiculo' not in df_veiculo_ajustado.columns:
        df_veiculo_ajustado['categoria_veiculo'] = None

    df_veiculo_ajustado.to_sql('dim_veiculo', con=conexao, if_exists='append', index=False)
    print("✓ Dados da 'dim_veiculo' inseridos com sucesso!")

    # =========================================================================
    # 4. Ajustar e Inserir: dim_local
    # =========================================================================
    # Nota: Como seu DF tem 'complemento' e o banco do diagrama original pedia 'latitude' e 'longitude',
    # se você quiser seguir estritamente o banco anterior, adaptamos aqui:
    df_local_ajustado = dim_local.copy()
    if 'latitude' not in df_local_ajustado.columns:
        df_local_ajustado['latitude'] = None
    if 'longitude' not in df_local_ajustado.columns:
        df_local_ajustado['longitude'] = None

    # Mantemos apenas as colunas que o banco possui
    df_local_ajustado = df_local_ajustado[['sk_local', 'bairro', 'endereco', 'latitude', 'longitude']]

    df_local_ajustado.to_sql('dim_local', con=conexao, if_exists='append', index=False)
    print("✓ Dados da 'dim_local' inseridos com sucesso!")

    # =========================================================================
    # 5. Ajustar e Inserir: dim_tempo
    # =========================================================================
    df_tempo_ajustado = dim_tempo.rename(columns={
        'data': 'data_acidente',
        'hora': 'hora_acidente',
        'nome_dia_semana': 'dia_semana',
        'is_dia_semana': 'fim_de_semana'
    })
    # Convertendo o booleano 'is_dia_semana' para texto Sim/Não se preferir
    df_tempo_ajustado['fim_de_semana'] = df_tempo_ajustado['fim_de_semana'].map({True: 'Não', False: 'Sim'})

    df_tempo_ajustado.to_sql('dim_tempo', con=conexao, if_exists='append', index=False)
    print("✓ Dados da 'dim_tempo' inseridos com sucesso!")

    # =========================================================================
    # 6. Ajustar e Inserir: fato_acidente
    # =========================================================================
    # O banco do diagrama original pedia 'quantidade_veiculos'. Se não houver no seu DF, criamos como nulo.
    df_fato_ajustado = fato_acidente.copy()
    if 'quantidade_veiculos' not in df_fato_ajustado.columns:
        df_fato_ajustado['quantidade_veiculos'] = None

    # Como você deixou latitude/longitude na fato no Pandas, mas o banco original não tinha elas lá,
    # filtramos para enviar apenas as colunas que a tabela fato do banco aceita:
    colunas_banco_fato = [
        'sk_acidente', 'sk_tempo', 'sk_local', 'sk_tipo_acidente',
        'sk_veiculo', 'quantidade_vitimas', 'quantidade_veiculos'
    ]
    df_fato_ajustado = df_fato_ajustado[[col for col in colunas_banco_fato if col in df_fato_ajustado.columns]]

    df_fato_ajustado.to_sql('fato_acidente', con=conexao, if_exists='append', index=False)
    print("✓ Dados da 'fato_acidente' inseridos com sucesso!")

    print("\n[SUCESSO] Todo o Data Warehouse foi populado corretamente!")

except Exception as e:
    print(f"\n[ERRO] Falha ao inserir dados no banco: {e}")

finally:
    # Fecha a conexão com o banco de dados de forma segura
    conexao.close()
    print("Conexão com o SQLite encerrada.")
```
