```python

import pandas as pd
import numpy as np

df_2014 = pd.read_csv('./data/2014/Acidentes-2014.csv', sep=';', encoding='utf-8')
df_2015 = pd.read_csv('./data/2015/acidentes_2015.csv', sep=';', encoding='utf-8')
df_2016 = pd.read_csv('./data/2016/acidentes_2016.csv', sep=';', encoding='utf-8')

df_janeiro_2015 = pd.read_csv('./data/2015/acidentes_janeiro_2015.csv', sep=',', encoding='utf-8')
df_fevereiro_2015 = pd.read_csv('./data/2015/acidentes_fevereiro_2015.csv', sep=',', encoding='utf-8')
df_marco_2015 = pd.read_csv('./data/2015/acidentes_marco_2015.csv', sep=',', encoding='utf-8')
df_abril_2015 = pd.read_csv('./data/2015/acidentes_abril_2015.csv', sep=',', encoding='utf-8')
df_maio_2015 = pd.read_csv('./data/2015/acidentes_maio_2015.csv', sep=',', encoding='utf-8')
print("Bases carregadas com sucesso!")
```

```python
# Imprimindo a lista exata de colunas de cada ano para comparação
print("--- COLUNAS DE 2014 ---")
print(df_2014.columns.tolist())

print("--- COLUNAS JANEIRO 2015 ---")
print(df_janeiro_2015.columns.tolist())

print("--- COLUNAS FEVEREIRO 2015 ---")
print(df_fevereiro_2015.columns.tolist())

print("--- COLUNAS MARÇO 2015 ---")
print(df_marco_2015.columns.tolist())

print("--- COLUNAS ABRIL 2015 ---")
print(df_abril_2015.columns.tolist())

print("--- COLUNAS MAIO 2015 ---")
print(df_maio_2015.columns.tolist())

print("\n--- COLUNAS DE 2015 ---")
print(df_2015.columns.tolist())

print("\n--- COLUNAS DE 2016 ---")
print(df_2016.columns.tolist())
```

```python
map_2014 = {'tipo': 'tipo_raw', 'detalhes': 'descricao', 'data': 'data'}
map_2015 = {'tipo_ocorrencia': 'ocorrencia_raw', 'tipo': 'tipo_raw', 'descricao': 'descricao', 'data_abertura': 'data', 'hora_abertura': 'hora'}
map_2016 = {'tipo de ocorrencia': 'ocorrencia_raw', 'tipo': 'tipo_raw', 'descricao': 'descricao', 'data de abertura': 'data', 'hora de abertura': 'hora', 'quantidade de vitimas': 'quantidade_vitimas'}
map_mensal = {'tipo': 'tipo_raw', 'detalhes': 'descricao', 'data': 'data'}

colunas_desejadas = ['data', 'hora', 'bairro', 'endereco', 'latitude', 'longitude', 'ocorrencia_raw', 'tipo_raw', 'descricao', 'quantidade_vitimas']

def processar_arquivo_mensal(df):
    df.columns = df.columns.str.lower().str.strip()
    df = df.rename(columns=map_mensal)
    colunas_presentes = [c for c in colunas_desejadas if c in df.columns]
    return df[colunas_presentes]

df_2014 = df_2014.rename(columns=map_2014)
df_2015 = df_2015.rename(columns=map_2015)
df_2016 = df_2016.rename(columns=map_2016)

df_janeiro_2015 = processar_arquivo_mensal(df_janeiro_2015)
df_fevereiro_2015 = processar_arquivo_mensal(df_fevereiro_2015)
df_marco_2015 = processar_arquivo_mensal(df_marco_2015)
df_abril_2015 = processar_arquivo_mensal(df_abril_2015)
df_maio_2015 = processar_arquivo_mensal(df_maio_2015)

df_consolidado = pd.concat([df_2014, df_2015, df_2016, df_janeiro_2015, df_fevereiro_2015, df_marco_2015, df_abril_2015, df_maio_2015], ignore_index=True)
display(df_consolidado.head())
```

```python

# Visualizando o tamanho da nova base e as primeiras linhas
print(f"Total de registros consolidados: {df_consolidado.shape[0]} linhas e {df_consolidado.shape[1]} colunas")
display(df_consolidado.head())
```

```python
colunas_texto = ['hora', 'bairro', 'endereco', 'descricao', 'ocorrencia_raw', 'tipo_raw']
df_consolidado[colunas_texto] = df_consolidado[colunas_texto].fillna('NÃO INFORMADO')
df_consolidado['quantidade_vitimas'] = pd.to_numeric(df_consolidado['quantidade_vitimas'], errors='coerce').fillna(1).astype(int)

df_consolidado['data'] = pd.to_datetime(df_consolidado['data'], errors='coerce')
df_consolidado = df_consolidado.dropna(subset=['data'])

for col in colunas_texto:
    df_consolidado[col] = df_consolidado[col].str.upper().str.strip()
```

```python
def classificar_veiculo(tipo):
    if 'MOTO' in tipo: return 'Motocicleta'
    if 'CICLOMOT' in tipo: return 'Ciclomotor'
    if 'AUTOM' in tipo or 'COLIS' in tipo: return 'Automóvel'
    if 'CICLISTA' in tipo: return 'Ciclista'
    if 'PEDESTRE' in tipo or 'ATROPELAMENTO' in tipo: return 'Pedestre'
    return 'Outros'

def categorizar_veiculo(tipo_veiculo):
    if tipo_veiculo in ['Motocicleta', 'Ciclomotor']: return 'Motorizado de duas rodas'
    if tipo_veiculo == 'Automóvel': return 'Motorizado de quatro ou mais rodas'
    if tipo_veiculo == 'Ciclista': return 'Não motorizado'
    if tipo_veiculo == 'Pedestre': return 'Pedestre'
    return 'Outros'

def classificar_ocorrencia(ocorrencia):
    if any(x in ocorrencia for x in ['COLISÃO', 'CHOQUE', 'ENGAVETAMENTO']): return 'Colisão'
    if 'ATROPELAMENTO ANIMAL' in ocorrencia: return 'Atropelamento de animal'
    if 'ATROPELAMENTO' in ocorrencia: return 'Atropelamento'
    if 'CAPOTAMENTO' in ocorrencia or 'TOMBAMENTO' in ocorrencia: return 'Perda de controle'
    return 'Outros'

def classificar_causa(row):
    oco = row['ocorrencia_raw']
    tipo = row['tipo_raw']
    desc = row['descricao']
    if 'ATROPELAMENTO ANIMAL' in oco or 'ANIMAL' in desc: return 'Atropelamento de animal'
    if 'ATROPELAMENTO' in oco or 'PEDESTRE' in tipo: return 'Atropelamento de pedestre'
    if 'CAPOTAMENTO' in oco or 'TOMBAMENTO' in oco: return 'Capotamento/Tombamento'
    if 'COLIS' in oco or 'CHOQUE' in oco or 'ENGAVET' in oco or 'COLIS' in tipo: return 'Colisão entre veículos'
    if 'CICLISTA' in tipo: return 'Acidente com ciclista'
    return 'Não classificado'

df_consolidado['tipo_veiculo'] = df_consolidado['tipo_raw'].apply(classificar_veiculo)
df_consolidado['categoria_veiculo'] = df_consolidado['tipo_veiculo'].apply(categorizar_veiculo)
df_consolidado['classificacao_ocorrencia'] = df_consolidado['ocorrencia_raw'].apply(classificar_ocorrencia)
df_consolidado['causa_acidente'] = df_consolidado.apply(classificar_causa, axis=1)

dim_tipo_acidente = df_consolidado[['causa_acidente', 'descricao']].drop_duplicates().reset_index(drop=True)
dim_tipo_acidente = dim_tipo_acidente.rename(columns={'descricao': 'descricao_detalhada'})
dim_tipo_acidente.insert(0, 'sk_tipo_acidente', dim_tipo_acidente.index + 1)

dim_veiculo = df_consolidado[['tipo_veiculo', 'categoria_veiculo']].drop_duplicates().reset_index(drop=True)
dim_veiculo.insert(0, 'sk_veiculo', dim_veiculo.index + 1)

dim_local = df_consolidado[['bairro', 'endereco', 'latitude', 'longitude']].drop_duplicates().reset_index(drop=True)
dim_local.insert(0, 'sk_local', dim_local.index + 1)

dim_tempo = df_consolidado[['data', 'hora']].drop_duplicates().reset_index(drop=True)
dim_tempo.insert(0, 'sk_tempo', dim_tempo.index + 1)
dim_tempo['ano'] = dim_tempo['data'].dt.year
dim_tempo['mes'] = dim_tempo['data'].dt.month
dim_tempo['dia'] = dim_tempo['data'].dt.day
dim_tempo['dia_semana'] = dim_tempo['data'].dt.day_name()
dim_tempo['fim_de_semana'] = np.where(dim_tempo['data'].dt.dayofweek.isin([5, 6]), 'Sim', 'Não')
dim_tempo['data_acidente'] = dim_tempo['data'].dt.strftime('%Y-%m-%d')

dim_ocorrencia = df_consolidado[['ocorrencia_raw', 'classificacao_ocorrencia']].drop_duplicates().reset_index(drop=True)
dim_ocorrencia = dim_ocorrencia.rename(columns={'ocorrencia_raw': 'tipo_ocorrencia'})
dim_ocorrencia.insert(0, 'sk_ocorrencia', dim_ocorrencia.index + 1)

display(dim_tempo.head())
```

```python
df_consolidado['data_str'] = df_consolidado['data'].dt.strftime('%Y-%m-%d')

fato_acidente = df_consolidado.merge(dim_tipo_acidente, left_on=['causa_acidente', 'descricao'], right_on=['causa_acidente', 'descricao_detalhada'], how='left')
fato_acidente = fato_acidente.merge(dim_veiculo, on=['tipo_veiculo', 'categoria_veiculo'], how='left')
fato_acidente = fato_acidente.merge(dim_local, on=['bairro', 'endereco', 'latitude', 'longitude'], how='left')
fato_acidente = fato_acidente.merge(dim_tempo, left_on=['data_str', 'hora'], right_on=['data_acidente', 'hora'], how='left')
fato_acidente = fato_acidente.merge(dim_ocorrencia, left_on='ocorrencia_raw', right_on='tipo_ocorrencia', how='left')

fato_acidente['quantidade_veiculos'] = 1

colunas_fato = ['sk_tempo', 'sk_local', 'sk_tipo_acidente', 'sk_ocorrencia', 'sk_veiculo', 'quantidade_vitimas', 'quantidade_veiculos']
fato_acidente = fato_acidente[colunas_fato]
fato_acidente.insert(0, 'sk_acidente', fato_acidente.index + 1)

display(fato_acidente.head())
```

```python
fato_acidente.to_csv('fato_acidente.csv', index=False, sep=';', encoding='utf-8')
dim_ocorrencia.to_csv('dim_ocorrencia.csv', index=False, sep=';', encoding='utf-8')
dim_veiculo.to_csv('dim_veiculo.csv', index=False, sep=';', encoding='utf-8')
dim_tipo_acidente.to_csv('dim_tipo_acidente.csv', index=False, sep=';', encoding='utf-8')
dim_local.to_csv('dim_local.csv', index=False, sep=';', encoding='utf-8')
dim_tempo.to_csv('dim_tempo.csv', index=False, sep=';', encoding='utf-8')
```

```python
import sqlite3
from sqlite3 import Error

def criar_tabelas_sqlite():
    # Nome do arquivo do banco de dados local
    # Se o arquivo não existir, o SQLite cria automaticamente na mesma pasta do script
    nome_banco = "dw_acidentes.db"

    # Scripts DDL para criação das tabelas
    #Dimensões primeiro, Fato por último por causa das Foreign Keys.
    tabelas_sql = {
        "dim_local": """
            CREATE TABLE IF NOT EXISTS dim_local (
                sk_local INTEGER PRIMARY KEY,
                bairro TEXT,
                endereco TEXT,
                latitude REAL,
                longitude REAL
            );
        """,
        "dim_tempo": """
            CREATE TABLE IF NOT EXISTS dim_tempo (
                sk_tempo INTEGER PRIMARY KEY,
                data_acidente TEXT, -- SQLite armazena datas como TEXT (YYYY-MM-DD)
                ano INTEGER,
                mes INTEGER,
                dia INTEGER,
                hora_acidente TEXT, -- SQLite não tem tipo TIME nativo, usa-se TEXT (HH:MM:SS)
                dia_semana TEXT,
                fim_de_semana TEXT
            );
        """,
        "dim_tipo_acidente": """
            CREATE TABLE IF NOT EXISTS dim_tipo_acidente (
                sk_tipo_acidente INTEGER PRIMARY KEY,
                causa_acidente TEXT,
                descricao_detalhada TEXT
            );
        """,
        "dim_ocorrencia": """
            CREATE TABLE IF NOT EXISTS dim_ocorrencia (
                sk_ocorrencia INTEGER PRIMARY KEY,
                tipo_ocorrencia TEXT,
                classificacao_ocorrencia TEXT
            );
        """,
        "dim_veiculo": """
            CREATE TABLE IF NOT EXISTS dim_veiculo (
                sk_veiculo INTEGER PRIMARY KEY,
                tipo_veiculo TEXT,
                categoria_veiculo TEXT
            );
        """,
        "fato_acidente": """
            CREATE TABLE IF NOT EXISTS fato_acidente (
                sk_acidente INTEGER PRIMARY KEY,
                sk_tempo INTEGER,
                sk_local INTEGER,
                sk_tipo_acidente INTEGER,
                sk_ocorrencia INTEGER,
                sk_veiculo INTEGER,
                quantidade_vitimas INTEGER,
                quantidade_veiculos INTEGER,
                FOREIGN KEY (sk_tempo) REFERENCES dim_tempo(sk_tempo),
                FOREIGN KEY (sk_local) REFERENCES dim_local(sk_local),
                FOREIGN KEY (sk_tipo_acidente) REFERENCES dim_tipo_acidente(sk_tipo_acidente),
                FOREIGN KEY (sk_ocorrencia) REFERENCES dim_ocorrencia(sk_ocorrencia),
                FOREIGN KEY (sk_veiculo) REFERENCES dim_veiculo(sk_veiculo)
            );
        """
    }

    try:
        print(f"Conectando/Criando o banco de dados local '{nome_banco}'...")
        # Abre a conexão com o arquivo
        conexao = sqlite3.connect(nome_banco)
        cursor = conexao.cursor()

        cursor.execute("PRAGMA foreign_keys = ON;")

        print("Conexão bem-sucedida! Criando tabelas...\n")

        # Iterando sobre o dicionário e executando os comandos SQL
        for nome_tabela, ddl in tabelas_sql.items():
            print(f"Criando a tabela: {nome_tabela}...")
            cursor.execute(ddl)
            print(f"Tabela {nome_tabela} configurada com sucesso!\n")

        # Salva as alterações no arquivo
        conexao.commit()
        print("Processo finalizado com sucesso. O arquivo de banco de dados está pronto para uso.")

    except Error as e:
        print(f"Erro ao acessar o SQLite: {e}")
    finally:
        # Fecha a conexão com o arquivo de forma segura
        if conexao:
            cursor.close()
            conexao.close()
            print("Conexão com o SQLite encerrada.")

if __name__ == "__main__":
    criar_tabelas_sqlite()
```

```python
import sqlite3

# 1. Criar ou abrir a conexão com o mesmo banco de dados SQLite
nome_banco = "dw_acidentes.db"
conexao = sqlite3.connect(nome_banco)

try:
    print("Iniciando a carga dos dados no SQLite...\n")

    # =========================================================================
    # 2. Ajustar e Inserir: dim_tipo_acidente
    # =========================================================================
    df_tipo_ajustado = dim_tipo_acidente.rename(columns={'tipo_acidente': 'causa_acidente'})
    # Criamos uma coluna vazia para a 'descricao_detalhada' caso não exista
    if 'descricao_detalhada' not in df_tipo_ajustado.columns:
        df_tipo_ajustado['descricao_detalhada'] = None

    df_tipo_ajustado.to_sql('dim_tipo_acidente', con=conexao, if_exists='append', index=False)
    print("✓ Dados da 'dim_tipo_acidente' inseridos com sucesso!")

    # =========================================================================
    # 3. Ajustar e Inserir: dim_veiculo
    # =========================================================================
    df_veiculo_ajustado = dim_veiculo.rename(columns={'veiculo': 'tipo_veiculo'})
    if 'categoria_veiculo' not in df_veiculo_ajustado.columns:
        df_veiculo_ajustado['categoria_veiculo'] = None

    df_veiculo_ajustado.to_sql('dim_veiculo', con=conexao, if_exists='append', index=False)
    print("✓ Dados da 'dim_veiculo' inseridos com sucesso!")

    # =========================================================================
    # 4. Ajustar e Inserir: dim_local
    # =========================================================================
    df_local_ajustado = dim_local.copy()
    if 'latitude' not in df_local_ajustado.columns:
        df_local_ajustado['latitude'] = None
    if 'longitude' not in df_local_ajustado.columns:
        df_local_ajustado['longitude'] = None

    # Mantem apenas as colunas que o banco possui
    df_local_ajustado = df_local_ajustado[['sk_local', 'bairro', 'endereco', 'latitude', 'longitude']]

    df_local_ajustado.to_sql('dim_local', con=conexao, if_exists='append', index=False)
    print("✓ Dados da 'dim_local' inseridos com sucesso!")

    # =========================================================================
    # 5. Ajustar e Inserir: dim_tempo
    # =========================================================================
    df_tempo_ajustado = dim_tempo.rename(columns={
        'data': 'data_acidente',
        'hora': 'hora_acidente',
        'nome_dia_semana': 'dia_semana',
        'is_dia_semana': 'fim_de_semana'
    })
    # Convertendo o booleano 'is_dia_semana' para texto Sim/Não
    df_tempo_ajustado['fim_de_semana'] = df_tempo_ajustado['fim_de_semana'].map({True: 'Não', False: 'Sim'})

    df_tempo_ajustado.to_sql('dim_tempo', con=conexao, if_exists='append', index=False)
    print("✓ Dados da 'dim_tempo' inseridos com sucesso!")

    # =========================================================================
    # 6. Ajustar e Inserir: fato_acidente
    # =========================================================================
    # Se não houver no seu cria como nulo.
    df_fato_ajustado = fato_acidente.copy()
    if 'quantidade_veiculos' not in df_fato_ajustado.columns:
        df_fato_ajustado['quantidade_veiculos'] = None

    colunas_banco_fato = [
        'sk_acidente', 'sk_tempo', 'sk_local', 'sk_tipo_acidente',
        'sk_veiculo', 'quantidade_vitimas', 'quantidade_veiculos'
    ]
    df_fato_ajustado = df_fato_ajustado[[col for col in colunas_banco_fato if col in df_fato_ajustado.columns]]

    df_fato_ajustado.to_sql('fato_acidente', con=conexao, if_exists='append', index=False)
    print("✓ Dados da 'fato_acidente' inseridos com sucesso!")

    print("\n[SUCESSO] Todo o Data Warehouse foi populado corretamente!")

except Exception as e:
    print(f"\n[ERRO] Falha ao inserir dados no banco: {e}")

finally:
    # Fecha a conexão com o banco de dados de forma segura
    conexao.close()
    print("Conexão com o SQLite encerrada.")
```
